```python
import numpy as np
import pandas as pd
```

```python
df = pd.DataFrame(
    [
        [8,8,4],    # row 1
        [7,9,5],    # row 2
        [6,10,6],   # row 3
        [5,12,7]    # row 4
    ],
    columns = ['cgpa', 'profile_score', 'lpa']
)
```

```python
import tensorflow
from tensorflow import keras
from keras import Sequential
from keras.layers import Dense
```

```python
model = Sequential()

model.add(
    # Create 2 neurons in this layer
    # activation funcition as linear
    # Each input sample contains 2 features
    Dense(2, activation='linear', input_dim = 2)
)

model.add(
    # Create 1 neurons for output layer
    # activation funcition as linear
    # Layer 1 output feed to this layer
    Dense(1, activation='linear')
)
```

```python
model.summary()
```

```python
model.get_weights()
```

```python
new_weights = [
    # W1
    np.array(
        [
            [0.1, 0.1],
            [0.1, 0.1]
        ],
        dtype = np.float32
    ),
    # b1
    np.array(
        [0., 0.],
        dtype = np.float32
    ),
    # W2
    np.array(
        [
            [0.1],
            [0.1]
        ],
        dtype = np.float32
    ),
    # b2
    np.array(
        [0.],
        dtype = np.float32
    ),
]
```

```python
model.set_weights(new_weights)
```

```python
model.get_weights()
```

```python
optimizer = keras.optimizers.Adam(learning_rate = 0.001)
model.compile(loss='mean_squared_error',optimizer = optimizer)
```

```python
model.fit(
    df.iloc[:, 0:-1].values, 
    df['lpa'].values,
    epochs = 100,
    verbose = 1,
    batch_size = 1
)
```
