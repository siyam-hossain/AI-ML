**Inefficient Code**

```python
import time
def fib(n):
    if n==1 or n==0:
        return 1
    else:
        return fib(n-1) + fib(n-2)
```

```python
fib(5)
```

```python
fib(40)
```

```python
start = time.time()
fib(36)

print(time.time()-start)
```

**Improved code**

```python
import time
def fibo(n, d):
    if n in d:
        return d[n]
    else:
        d[n] = fibo(n-1, d) + fibo(n-2, d)
        return d[n]
```

```python
d = {0 : 1, 1 : 1}

fibo(38, d)
```

```python
start = time.time()

print(f"50 : {fibo(50, d)}")

print(f"time: {time.time()-start}")
```
