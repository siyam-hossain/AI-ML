
### `make_classification()` — Parameters

Generates synthetic data for classification using `sklearn.datasets`.

- `X`: Input features
- `y`: Target class labels
- `numpy`: Numerical operations and arrays

| Parameter | Value | Description |
|---|---:|---|
| `n_samples` | 100 | Number of samples |
| `n_features` | 2 | Number of features per sample |
| `n_informative` | 1 | Features useful for classification |
| `n_redundant` | 0 | Number of redundant features |
| `n_classes` | 2 | Number of classes (0 and 1) |
| `n_clusters_per_class` | 1 | Clusters per class |
| `random_state` | 41 | Ensures reproducible data |
| `hypercube` | False | Changes cluster-center generation |
| `class_sep` | 15 | Controls separation between classes |

**Output:**
- `X.shape` → `(100, 2)` — 100 samples, 2 features.
- `y.shape` → `(100,)` — 100 class labels.

**Use:** Generate data to train and visualize a Perceptron classifier.

```python
from sklearn.datasets import make_classification
import numpy as np
X, y = make_classification(
    n_samples = 100,
    n_features = 2,
    n_informative = 1,
    n_redundant = 0,
    n_classes = 2,
    n_clusters_per_class = 1,
    random_state = 41,
    hypercube = False,
    class_sep = 15
)
```

```python
print(X.shape)
print(X)
```

```python
print(y.shape)
print(y)
```

```python
import matplotlib.pyplot as plt
plt.figure(figsize=(10, 6))
plt.scatter(
    X[ : , 0],
    X[ : , 1],
    c = y,
    cmap = 'bwr',
    s = 100
)
plt.show()
```

```python
def perceptron(x, y): 

    w1 = w2 = b = 1
    lr = 0.1

    epochs = 1000

    for j in range(epochs):
        for i in range(X.shape[0]):

            z = w1 * X[i][0] + w2 * X[i][1] + b

            if z*y[i] < 0: 
                w1 = w1 + lr * y[i] * X[i][0]
                w2 = w2 + lr * y[i] * X[i][1]
                b = b + lr * y[i]
    return w1, w2, b
```

```python
w1, w2, b = perceptron(X, y)
```

```python
print(f"w1: {w1}, w2: {w2}, b: {b}")
```

```python
m = -(w1 / w2)
c = -(b / w2)

print(f"m: {m}, c: {c}")
```

```python
x_input = np.linspace(-3, 3, 100)
y_input = m * x_input + c

plt.figure(figsize=(10, 6))
plt.plot(
    x_input,
    y_input,
    color = 'red',
    linewidth = 3
)

plt.scatter(
    X[ : , 0],
    X[ : , 1],
    c = y,
    cmap = 'bwr',
    s = 100
)
plt.show()
```
